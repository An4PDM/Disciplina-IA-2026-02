19. Crie dois arrays de numeros inteiros aleatorios de tamanho 20 e calcule o
angulo entre eles (produto escalar normalizado).

```python
a = np.random.randint(1, 20, size=20)
b = np.random.randint(1, 20, size=20)

# O np.dot permite multiplicar os elementos que estão na mesma posição e depois soma tudo (produto escalar entre dois vetores).
# O np.linalg.noem permite calcular o tamanho do vetor.
# Portanto, a linha abaixo basicamente corresponte ao calculo do angulo entre os dois vetores: cos(θ) = produto escalar / (tamanho de A × tamanho de B)
cos_angulo = np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

# Definimos o angulo como radianos
angulo_rad = np.arccos(cos_angulo)

# Por fim, o angulo em graus
angulo_graus = np.degrees(angulo_rad)

print(a, b)
print(cos_angulo)
print(angulo_rad)
print(angulo_graus)
```

Justificativa: foi calculado o produto escalar entre os dois arrays com np.dot e dividido pelo produto das normas dos vetores, obtidas com np.linalg.norm(). Essa operação corresponde ao produto escalar normalizado, que resulta no cosseno do ângulo entre os dois vetores. Depois, foi utilizada a função np.arccos() para encontrar o ângulo correspondente em radianos. Por fim, np.degrees() foi utilizada para converter o resultado de radianos para graus, obtendo assim o ângulo entre os dois arrays.

# Exercícios de Scipy

1. Calcule a integral da funcao f(x) = x² sin(x) no intervalo de 0 a π.

```python
from scipy import integrate

# Função python para  f(x) = x² sin(x)
def f(x):
    return x**2 * np.sin(x)

# integrate.quad faz integração numérica (quadratura) entre os limites 0 e π
# retorna o valor da integral e uma estimativa do erro
resultado, erro_estimado = integrate.quad(f, 0, np.pi)

print(resultado)
print(erro_estimado)
```

Justificativa: Foi usado o módulo integrate para realizar o cálculo numérico da integral definida. A função f(x) representa matematicamente a expressão f(x)=x² sin(x), utilizando np.sin(x) para calcular o seno de x.

A função integrate.quad() foi escolhida porque permite calcular numericamente uma integral definida informando a função e os limites de integração. Nesse caso, os limites são 0 e np.pi

2. Resolva a equacao diferencial dy/dx = x−y, com y(0) = 1, no intervalo de 0
a 10.

```python
from scipy.integrate import solve_ivp
import numpy as np

# define a equação diferencial dy/dx = x - y
def dy_dx(x, y):
    return x - y

# solve_ivp resolve numericamente a EDO
# [0, 10] é o intervalo de x, [1] é a condição inicial y(0)=1
# t_eval define em quais pontos queremos ver o resultado
# O np.linspace define 100 pontos entre 0 e 10
sol = solve_ivp(dy_dx, [0, 10], [1], t_eval=np.linspace(0, 10, 100))

print("x:", sol.t) # valores de x
print("y:", sol.y[0]) # valores correspondentes de y
```

Justificativa: A função solve_ivp de scipy foi usada para resolver numericamente uma equação diferencial. Primeiro, é definida a equação dy/dx = x - y. Depois, "solve_ivp" calcula uma aproximação da solução no intervalo de 0 a 10, considerando y(0) = 1 como condição inicial. O "np.linspace" cria 100 pontos igualmente espaçados nesse intervalo, permitindo visualizar os resultados de forma mais detalhada. Por fim, o código mostra os valores de x e os valores aproximados de y encontrados.

4. Use a funcao de otimizacao de SciPy para maximizar f(x) = −x²+4x+1.

```python
from scipy.optimize import minimize_scalar

# scipy só tem função pronta pra minimizar, então maximizar ficaria como f(x) = minimizar - f(x)
f_negativa = lambda x: -(-x**2 + 4*x + 1)
resultado = minimize_scalar(f_negativa)

x_max = resultado.x
f_max = -resultado.fun  # desfaz o sinal negativo pra voltar ao valor real

print("x que maximiza:", x_max)
print("Valor máximo:", f_max)
```

Justificativa: O código utiliza a função minimize_scalar do SciPy para encontrar o valor máximo da função. Como essa função serve para encontrar mínimos, a função original é multiplicada por -1, transformando o problema de maximização em minimização. Depois, o sinal negativo é desfeito para obter novamente o valor máximo real. Por fim, o código mostra o valor de x onde ocorre o máximo e o valor máximo encontrado.

5. Calcule a transformada de Fourier de um sinal composto de sin(2πt) e
cos(4πt).

```python
from scipy.fft import fft, fftfreq
import numpy as np

fs = 1000  # taxa de amostragem (quantos pontos por segundo)
t = np.linspace(0, 1, fs, endpoint=False)  # eixo do tempo, 1 segundo de sinal
sinal = np.sin(2*np.pi*t) + np.cos(4*np.pi*t)  # sinal combinado

# fft converte o sinal do domínio do tempo pro domínio da frequência
transformada = fft(sinal)
# fftfreq calcula quais frequências correspondem a cada posição do resultado da fft
frequencias = fftfreq(len(t), 1/fs)

print("Frequências:", frequencias[:10])
print("Magnitude:", np.abs(transformada)[:10])
```

Justificativa: O código utiliza a Transformada Rápida de Fourier (FFT) para analisar as frequências presentes em um sinal. Primeiro, é criado um sinal combinando uma função seno e uma função cosseno. A função fft transforma esse sinal do domínio do tempo para o domínio da frequência, permitindo identificar quais frequências estão presentes. Já fftfreq calcula as frequências correspondentes aos resultados da transformação. Depois, np.abs calcula a magnitude dessas frequências, mostrando a intensidade de cada componente do sinal.

7. Use a funcao de interpolacao cubica de SciPy para interpolar os dados
(0, 0), (1,2),(2,3),(3,5),(4,4).

```python
from scipy.interpolate import interp1d
import numpy as np

x = np.array([0, 1, 2, 3, 4])
y = np.array([0, 2, 3, 5, 4])

# kind='cubic' ajusta uma curva suave (polinômio de grau 3) passando pelos pontos
interpolador = interp1d(x, y, kind='cubic')

# gera 50 pontos novos dentro do intervalo original pra "preencher" a curva
x_novo = np.linspace(0, 4, 50)
y_novo = interpolador(x_novo)

print("Valores interpolados:", y_novo)
```

Por fim, o código exibe os valores interpolados.

8. Resolva o sistema de equacoes 3x + 2y = 5 e x+4y = 6.

```python
import numpy as np
from scipy.linalg import solve

# Matriz dos coeficientes (3x,2y,x,4y).
A = np.array([[3, 2],
              [1, 4]])

# Valores do lado direito das equações (depois do sinal de =)
b = np.array([5, 6])

# O "solve" resolve o sistema A * x = b
solucao = solve(A, b)

print("x =", solucao[0])
print("y =", solucao[1])
```

9. Ajuste uma curva quadratica aos pontos (1,2),(2,4),(3,5),(4,7),(5,8).

```python
from scipy.optimize import curve_fit
import numpy as np

# modelo que queremos ajustar aos dados: uma parábola
def quadratica(x, a, b, c):
    return a*x**2 + b*x + c

x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 5, 7, 8])

# curve_fit encontra os melhores valores de a, b, c que aproximam a curva dos pontos
parametros, _ = curve_fit(quadratica, x, y)
a, b, c = parametros

print(f"a={a}, b={b}, c={c}")
```

Justificativa: O código utiliza a função curve_fit do SciPy para encontrar uma curva que melhor se ajusta aos pontos fornecidos. Primeiro, é definida uma função quadrática, que representa uma parábola. Depois, curve_fit calcula os melhores valores de a, b e c para que essa parábola fique o mais próxima possível dos valores de x e y. Por fim, os valores encontrados para esses parâmetros são exibidos.

10. Calcule a integral dupla da funcao f(x,y) = xy + x² no intervalo [0,2]
para x e [0,1] para y.

```python
from scipy import integrate

# Define a função que será integrada.
# O scipy usa a ordem (y, x) para a função na integral dupla.
def f(y, x):
    return x * y + x**2

# Calcula a integral dupla usando dblquad.
# x varia de 0 até 2 e y varia de 0 até 1.
# A função retorna o resultado da integral e uma estimativa do erro.
resultado, erro = integrate.dblquad(f, 0, 2, 0, 1)

print(resultado)
```

Justificativa:
O código utiliza a função dblquad da biblioteca SciPy para calcular numericamente uma integral dupla. Primeiro, é definida a função f(x,y) = xy + x². Depois, dblquad calcula a integral considerando x no intervalo de 0 a 2 e y no intervalo de 0 a 1. O resultado da integral é armazenado na variável resultado, enquanto erro recebe uma estimativa do erro do cálculo. Por fim, o resultado é exibido na tela.

Um detalhe importante é que a função usada pelo dblquad deve receber os argumentos na ordem (y, x).

16. Resolva a equacao diferencial de segunda ordem d²y/dx² + y = 0, com condicoes iniciais y(0) = 1 e y′(0) = 0.

```python
from scipy.integrate import solve_ivp
import numpy as np

# solve_ivp só resolve EDOs de 1ª ordem, então quebramos a equação de 2ª ordem em duas de 1ª:
# y1 = y (a função original) e y2 = y' (a derivada dela)
def sistema(x, estado):
    y1, y2 = estado
    dy1dx = y2      # derivada de y1 é y2 (por definição)
    dy2dx = -y1      # derivada de y2 vem da equação original: y'' = -y
    return [dy1dx, dy2dx]

# condições iniciais: [y(0)=1, y'(0)=0]
sol = solve_ivp(sistema, [0, 10], [1, 0], t_eval=np.linspace(0, 10, 100))

print("x:", sol.t)
print("y:", sol.y[0])
print("y':", sol.y[1])
```

Justificativa: O código utiliza a função solve_ivp do SciPy para resolver numericamente uma equação diferencial de segunda ordem. Como solve_ivp trabalha com equações de primeira ordem, a equação é transformada em um sistema de duas equações, usando y1 para representar y e y2 para representar y'. As condições iniciais são y(0)=1 e y'(0)=0. Depois, solve_ivp calcula uma aproximação da solução no intervalo de 0 a 10. Por fim, o código exibe os valores de x, y e y' encontrados.

19. Realize a transformada de Fourier de um sinal discreto composto de senos
e cossenos.

```python
from scipy.fft import fft, fftfreq
import numpy as np

# Define a quantidade de pontos usados para representar o sinal.
n = 100

# Cria 100 pontos de tempo entre 0 e 1 segundo.
# endpoint=False evita incluir o valor final 1.
t = np.linspace(0, 1, n, endpoint=False)

# Cria um sinal formado pela combinação de duas ondas: uma de 5 Hz e outra de 10 Hz.
sinal = np.sin(2*np.pi*5*t) + np.cos(2*np.pi*10*t)

# Aplica a Transformada Rápida de Fourier (FFT) convertendo o sinal do domínio do tempo para o domínio da frequência.
transformada = fft(sinal)

# Calcula as frequências correspondentes aos valores da FFT.
frequencias = fftfreq(n, 1/n)

# Mostra as 10 primeiras frequências encontradas.
print("Frequências:", frequencias[:10])

# Mostra a magnitude das 10 primeiras frequências, indicando a intensidade de cada componente do sinal.
print("Magnitude:", np.abs(transformada)[:10])
```

Justificativa:
O código utiliza a Transformada Rápida de Fourier (FFT) para identificar as frequências presentes em um sinal. Primeiro, é criado um sinal formado pela combinação de duas ondas, uma de 5 Hz e outra de 10 Hz. A função fft transforma o sinal do domínio do tempo para o domínio da frequência, enquanto fftfreq identifica as frequências correspondentes. Por fim, np.abs calcula a magnitude das frequências, permitindo observar a intensidade de cada componente do sinal.

8. Plote um grafico de linha com intervalos de confianca (sombramento) para
a funcao y = sin(x).

```python
#  grade de pontos no eixo X
x = np.linspace(0, 2 * np.pi, 200)

# Calculo do valor central y e a margem do intervalo de confiança (desvio simulado)
y = np.sin(x)
desvio_erro = 0.15

# Criação do gráfico com a curva central e a região sombreada ao redor
plt.figure()
plt.plot(x, y, color="blue", linewidth=2, label="Estimativa Média (Seno)")

# fill_between preenche a área vertical de (y - erro) até (y + erro) para expressar o desvio
plt.fill_between(x, y - desvio_erro, y + desvio_erro, color="blue", alpha=0.15,
                 label="Intervalo de Confiança (95%)")

# Customização da grade e legendas
plt.xlabel("Fase (Radianos)")
plt.ylabel("Amplitude")
plt.legend(loc="upper right")

plt.show()
```

Justificativa: O código utiliza np.linspace() para criar 200 pontos entre 0 e 2π, permitindo representar a curva do seno com maior suavidade. A função np.sin() calcula os valores da função seno e desvio_erro define uma margem de erro simulada. plt.plot() desenha a curva central, enquanto plt.fill_between() cria uma região sombreada entre os limites inferior e superior do erro, facilitando a visualização da variação em torno da estimativa.

13. Gere um grafico de dispersao 3D para 100 pontos aleatorios utilizando
Matplotlib.

```python
# Coordenadas aleatórias
x = np.random.randn(100)
y = np.random.randn(100)
z = np.random.randn(100)

# Configuração do eixo com projeção espacial tridimensional
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection="3d")

# Plotagem dos pontos de dispersão espacial 3D
# O mapeamento de cores ajuda a guiar a percepção de profundidade tridimensional
grafico_3d = ax.scatter(x, y, z, c=z, cmap="coolwarm", s=50, edgecolors="black")

# barra de intensidade de profundidade de cores
fig.colorbar(grafico_3d, ax=ax, label="Eixo de Cores Z")

# 5. Adicionamos as legendas espaciais para cada um dos três eixos coordenados
ax.set_title("Gráfico de Dispersão 3D")
ax.set_xlabel("Eixo Espacial X")
ax.set_ylabel("Eixo Espacial Y")
ax.set_zlabel("Eixo Espacial Z")

plt.show()
```

Justificativa: O código utiliza np.random.randn() para gerar 100 valores aleatórios para cada um dos três eixos (X, Y e Z). add_subplot(..., projection="3d") cria um eixo tridimensional, permitindo representar os dados em três dimensões. A função ax.scatter() gera o gráfico de dispersão 3D, utilizando c=z para relacionar as cores aos valores do eixo Z e cmap para definir o gradiente de cores. colorbar() adiciona uma escala para interpretar essas cores. Por fim, set_xlabel(), set_ylabel() e set_zlabel() identificam os três eixos, facilitando a interpretação do gráfico.
