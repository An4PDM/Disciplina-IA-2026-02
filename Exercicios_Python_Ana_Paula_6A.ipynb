12. Contagem de Caracteres: Conte o número de ocorrências de cada caractere em uma string
fornecida

```python
palavra = input("Digite uma palavra: ")

palavra = list(palavra) # Transforma a palavra em uma lista
contador = {} # A chave será a letra e o valor será o numero de ocorrencias

for letra in palavra: # Itera cada item (letra)
  if letra in contador:
    # Adiciona +1 caso já exista
    contador[letra] +=1

  else:
    contador[letra] = 1

print(contador)
```

18. Número Primo: Crie um programa que verifique se um número fornecido é primo.

Como Descobrir se um Número é Primo:

Para saber se um número grande é primo, você testa a divisão dele pelos números primos menores (2, 3, 5, 7, etc.):
- Veja se ele é par (se for e for maior que 2, não é primo).
- Some os algarismos para ver se divide por 3.
- Veja se termina em 0 ou 5 para saber se divide por 5.
- Se a divisão exata não acontecer por nenhum primo menor cuja raiz quadrada seja menor ou igual ao número testado, ele é primo

```python
from math import sqrt

primos_menores = [2,3,5,7,11]

n = int(input("Insira um valor para descobrir se é primo: "))
raiz = sqrt(n) # Guarda a raíz desse número


for x in primos_menores:
  # Verifica se o valor fornecido é divisível por algum dos primos menores que a raiz desse valor fornecido.
  if x <= raiz:
    if n % x == 0:
      print("Não é um número primo.")
      break # Paramos o loop pois já não seria necessário verificar mais vezes

else:
  print("É um número primo.")
```

19. Jogo de Adivinhação: Implemente um jogo onde o usuário deve adivinhar um número gerado aleatoriamente pelo computador.

```python
import numpy as np

n_gerado = np.random.randint(1,10) # Geração de valor inteiro aleatório
tentativas = 1
print("Você terá 5 tentativas ao total.")
n_usuario = int(input("Descubra qual número foi gerado de 1 a 10: "))

while tentativas < 5: # O usuario terá, no maximo, 5 tentativas

  if n_gerado == n_usuario:
    print("Parabéns! Você acertou. ")
    break # Se acertar, o loop é parado.

  else:
    n_usuario = int(input("Tente novamente: "))

  tentativas += 1

print(f"O número aleatório gerado foi {n_gerado}.")
```

21. Busca Binária: Implemente a busca binária para encontrar um elemento em uma
lista ordenada.

Busca Binária: A lista deve estar em ordem crescente, o programa lê o número do meio e descarta a metade caso seja menor que o numero procurado.


```python
lista = [20,15,23,45,71,10,5]

n = int(input("Qual número será procurado? ")) # Elemento procurado

# Primeiro, ordenamos a lista
lista.sort()
print(lista)

i = lista[0:] # Inicio da lista
f = lista[:0] # Fim da lista

while i <= f:
# Verificação do índice do valor do meio
    posicao = len(lista) // 2 # Operador de divisão que "ignora" a casa decimal

    if lista[posicao] == n:
      print(f"O número procurado {n} foi encontrado.")
      break

    elif lista[posicao] < n:
      lista = lista[posicao:] # Eliminamos a metade da esquerda (numeros menores)
      continue

    elif lista[posicao] > n:
      lista = lista[:posicao] # Eliminamos a metade da direita (numeros maiores)
      continue
  else:
    print(f"O número {n} não está presente na lista.")



```

25. Classificação de Números: Classifique uma lista de números em três categorias: positivos,
negativos e zeros.

```python
lista = [-1,0,20,6,-12,4,0,-6]
positivos = []
negativos = []
zeros = []

for x in lista:
  if x == 0:
    zeros.append(x)

  elif x < 0:
    negativos.append(x)

  elif x > 0:
    positivos.append(x)

print("Lista com positivos: ", positivos)
print("Lista com negativos: ", negativos)
print("Lista com zeros: ", zeros)
```

29. Números FizzBuzz: Implemente o clássico problema FizzBuzz para números de 1 a 100.
- Número múltiplo de 3: Fizz;
- Número múltiplo de 5: Buzz;
- Número múltiplo de 3 e 5: FizzBuzz

```python
n = 1

while (n <= 100):

  if  n % 3 == 0 and n % 5 == 0:
    print("FizzBuzz!")

  elif n % 5 == 0:
    print("Buzz!")

  elif n % 3 == 0:
    print("Fizz!")

  else:
    print(n)

  n += 1
```

```python
palavra = input("Insira uma palavra: ")
vogais = ['a','e','i','o','u']
contador = 0

palavra = list(palavra)
for p in palavra:
  for v in vogais:
    if v == p:
      contador += 1

print(f"A palavra fornecida possui {contador} vogais.")
```
