18. Número Primo: Crie um programa que verifique se um número fornecido é primo.

Para saber se um número grande é primo, você testa a divisão dele pelos números primos menores (2, 3, 5, 7, 11).


```python
# Para evitar cálculo manual, foi importada a função sqrt que permite calcular a raiz quadrada de um valor fornecido automaticamente.
from math import sqrt


n = int(input("Insira um valor para descobrir se é primo: "))

# Calcula a raiz quadrada de n e armazena o resultado na variável raiz.
# Vamos usar esse valor para definir até onde precisamos testar os divisores.
raiz = sqrt(n)


# Lista contendo os números primos até 11 que serão usados para testar se n é divisível.
primos_menores = [2,3,5,7,11]

# Percorre cada número da lista de primos.
for x in primos_menores:

# Verifica se o primo atual é menor ou igual à raiz quadrada de n.
# Só precisamos testar divisores até a raiz quadrada de n.
  if x <= raiz:
    # Verifica se n é divisível por x.
    # O operador % retorna o resto da divisão.
    # Se o resto for 0, significa que n é divisível por x.
    if n % x == 0:
      print("Não é um número primo.")
      break # Paramos o loop pois já não seria necessário verificar mais vezes

# O else abaixo pertence ao for, ele é executado somente se o loop terminar e o break não for executado.
else:
  print("É um número primo.")


```

```python
matriz = [[4,16,12],[9,10,31],[34,2,45]]

# Lista que irá armazenar a matriz transposta
matriz_transposta = []

# Percorre as colunas da matriz original
# A quantidade de colunas será dada pelo tamanho da primeira linha, ou seja 3
for coluna in range(len(matriz[0])):

    # Cria uma nova linha para a matriz transposta
    nova_linha = []

    # Percorremos todas as linhas da matriz
    # Ao manter a coluna fixa e variar as linhas, pegamos os valores que formarão uma nova linha.
    for linha in range(len(matriz)):

        # Pega o elemento que está na linha atual
        # e na coluna que estamos percorrendo.
        nova_linha.append(matriz[linha][coluna])

    matriz_transposta.append(nova_linha)

print(matriz)
print("---------------------------------------")
print(matriz_transposta)
```

21. Busca Binária: Implemente a busca binária para encontrar um elemento em uma
lista ordenada.

Busca Binária: A lista deve estar em ordem crescente, o programa lê o número do meio e descarta a metade caso seja menor que o numero procurado.


```python
lista = [20, 15, 23, 45, 71, 10, 5]

n = int(input("Qual número será procurado? "))  # Elemento procurado

# Primeiro, ordenamos a lista
lista.sort()
print(lista)

# Define o início e o fim da parte da lista que será pesquisada
inicio = 0
fim = len(lista) - 1

# Enquanto ainda existir uma parte da lista para pesquisar
while inicio <= fim:

    # Calcula a posição do elemento do meio
    posicao = (inicio + fim) // 2

    # Verifica se o elemento do meio é o número procurado
    if lista[posicao] == n:
        print(f"O número procurado {n} foi encontrado.")
        break

    # Se o elemento do meio for menor que o procurado, descartamos a metade da esquerda
    elif lista[posicao] < n:
        inicio = posicao + 1

    # Se o elemento do meio for maior que o procurado, descartamos a metade da direita
    elif lista[posicao] > n:
        fim = posicao - 1

# Se o início ultrapassou o fim, então significa que o número não foi encontrado
else:
    print(f"O número {n} não está presente na lista.")
```

22. Criptografia de Cifra de César: Crie uma função para criptografar uma mensagem usando a cifra de
César.

```python
# O alfabeto foi armazenado em uma única string para facilitar na criação, pois seria mais trabalho criar como lista.
alfabeto = 'ABCDEFGHIJKLMNOPQRSTUVWXYZ'

# Aqui o deslocamento foi definido como 3, pois a maior parte dos exemplos vistos utilizavam esse deslocamento.
deslocamento = 3

# Lista que armazenará cada letra da string criptografada
msg_criptografada = []

msg = input("Qual a mensagem que deseja criptografar? ")

for x in msg:
  i = alfabeto.index(x) # Armazena o índice correspondente da letra no alfabeto

  # Para evitar o problema quando o deslocamento ultrapassar a letra Z, utilizamos o operador "%" com 26, que é a quantidade de letras do alfabeto.
  # O índice da letra é somado ao deslocamento e o resultado do resto da divisão por 26 garante que o novo índice permaneça entre 0 e 25.
  letra = alfabeto[(i + deslocamento) % 26]
  msg_criptografada.append(letra)

msg_criptografada = ''.join(msg_criptografada)
msg_criptografada
```

24. Validação de CPF: Implemente uma função para validar um CPF brasileiro.

```python
def validar_cpf(cpf):

    # Um CPF possui 11 dígitos.
    # Se tiver uma quantidade diferente, já sabemos que é inválido.
    if len(cpf) != 11:
        return False

    # Primeiro digito
    soma = 0

    # Para calcular o primeiro dígito, usamos apenas os 9 primeiros números.
    # Os pesos começam em 10 e diminuem até 2.
    for i in range(9):
        soma += int(cpf[i]) * (10 - i)

    # Pegamos o resto da divisão da soma por 11, pois ele será usado para encontrar o dígito verificador.
    resto = soma % 11

    # Se o resto for 0 ou 1, o dígito será 0.
    # Caso contrário, fazemos 11 menos o resto.
    if resto < 2:
        digito1 = 0
    else:
        digito1 = 11 - resto

    # Comparamos o dígito calculado com o 10º número do CPF,
    # que é o primeiro dígito verificador.
    if digito1 != int(cpf[9]):
        return False

    # Segundo dígito
    soma = 0

    # Para calcular o segundo dígito, usamos os 10 primeiros números, incluindo o primeiro dígito que acabamos de conferi.
    # Os pesos agora começam em 11 e diminuem até 2.
    for i in range(10):
        soma += int(cpf[i]) * (11 - i)

    # Denovo calculamos o resto da divisão por 11.
    resto = soma % 11

    # Aplicamos a mesma regra anterior para encontrar o segundo dígito
    if resto < 2:
        digito2 = 0
    else:
        digito2 = 11 - resto

    # Comparamos o dígito calculado com o último número do CPF.
    if digito2 != int(cpf[10]):
        return False

    # Se os dois dígitos estiverem corretos, o CPF é válido.
    return True


# Pede o CPF ao usuário
# O CPF deve ser digitado somente com números.
cpf = input("Digite o CPF (só numeros): ")

# Chama a função e verifica o resultado.
if validar_cpf(cpf):
    print("CPF válido!")
else:
    print("CPF inválido!")
```

JUSTIFICATIVA: Primeiro, é verificado se o CPF possui exatamente 11 dígitos, pois essa é a quantidade necessária para realizar a validação.

Para calcular o primeiro dígito verificador, são utilizados os 9 primeiros números do CPF, multiplicados por pesos que vão de 10 até 2. A soma desses resultados é dividida por 11, e o resto da divisão é utilizado para determinar o primeiro dígito. Esse dígito calculado é então comparado com o 10º número do CPF.

DepoIs, é realizado o cálculo do segundo dígito verificador, dessa vez utilizando os 10 primeiros números, incluindo o primeiro dígito verificador. Os pesos vão de 11 até 2, e o mesmo processo de divisão por 11 é aplicado.

Caso algum dos dígitos calculados seja diferente do dígito informado no CPF, a função retorna False, indicando que o CPF é inválido. Se os dois dígitos estiverem corretos, a função retorna True, indicando que o CPF passou pela validação.

25. Classificação de Números: Classifique uma lista de números em três categorias: positivos,
negativos e zeros.

```python
lista = [-1,0,20,6,-12,4,0,-6]
positivos = []
negativos = []
zeros = []

# O loop percorrerá a lista, definindo cada valor como "x".
for x in lista:
  # Caso "x" seja igual a 0, a lista "zeros" armazenará "x".
  if x == 0:
    zeros.append(x)

  # Caso "x" seja menor que 0, a lista "negativos" armazenará "x".
  elif x < 0:
    negativos.append(x)

  # Caso "x" seja maior que 0, a lista "positivos" armazenará "x".
  elif x > 0:
    positivos.append(x)

print("Lista com positivos: ", positivos)
print("Lista com negativos: ", negativos)
print("Lista com zeros: ", zeros)
```

- Número múltiplo de 3: Fizz;

- Número múltiplo de 3 e 5: FizzBuzz

30. Jogo da Forca: Desenvolva um jogo simples da forca onde o usuário deve adivinhar uma palavra.

```python
# Palavra que o usuário deverá descobrir
palavra = "abobora"

# Cria uma lista com "_" para esconder cada letra da palavra
descoberta = ["_"] * len(palavra)

# Define a quantidade de tentativas
tentativas = 6

print("===------- JOGO DA FORCA --------===")

# O jogo continua enquanto ainda houver tentativas
# e a palavra não tiver sido descoberta
while tentativas > 0 and "_" in descoberta:

    print("Palavra:", " ".join(descoberta))
    print("Tentativas restantes:", tentativas)

    # Solicita uma letra ao usuário
    letra = input("Digite uma letra: ").lower()

    # Verifica se a letra está presente na palavra
    if letra in palavra:

        # Percorre a palavra procurando a letra digitada
        for i in range(len(palavra)):

            # Se encontrar a letra, coloca ela na posição correspondente
            if palavra[i] == letra:
                descoberta[i] = letra

        print("Acertou!")

    else:
        # Se a letra não estiver na palavra, perde uma tentativa
        tentativas -= 1
        print("Você errou!")

# Verifica se o usuário conseguiu descobrir a palavra
if "_" not in descoberta:
    print("Parabéns! Você descobriu a palavra:", palavra)

else:
    print("Você perdeu!")
    print("A palavra era:", palavra)
```

JUSTIFICATIVA: A palavra é armazenada em uma variável e uma segunda lista é criada contendo _, representando as letras que ainda não foram descobertas. O while mantém o jogo em execução enquanto o jogador ainda possuir tentativas e houver letras escondidas na palavra.

A cada rodada, o usuário informa uma letra e o programa verifica, utilizando in, se ela pertence à palavra. Caso pertença, um for percorre as posições da palavra para encontrar todas as ocorrências daquela letra e substitui os _ correspondentes na lista descoberta. Caso a letra não esteja presente, uma tentativa é retirada.

Ao final, o programa verifica se ainda existem _ na lista. Se não houver, significa que todas as letras foram descobertas e o jogador venceu. Caso as tentativas acabem antes disso, o jogador perde e a palavra correta é exibida.

31. Ordenação por Bolha: Implemente o algoritmo de ordenação por bolha (Bubble Sort).

```python
lista = [5, 2, 8, 1, 3]

# Percorre a lista várias vezes para comparar os elementos
for i in range(len(lista)):

    # Compara cada elemento com o próximo
    # O -1 evita tentar acessar uma posição que não existe
    for j in range(len(lista) - 1):

        # Se o elemento atual for maior que o próximo,
        # significa que eles estão na ordem errada
        if lista[j] > lista[j + 1]:

            # Troca os dois elementos de posição
            lista[j], lista[j + 1] = lista[j + 1], lista[j]


# Exibe a lista ordenada
print(lista)
```

Justificativa: Nessa solução, utilizamos dois loops "for": o primeiro controla a quantidade de vezes que a lista será percorrida, enquanto o segundo percorre as posições da lista para realizar as comparações.

Usamos range(len(lista)) porque foi necessario trabalhar com os índices dos elementos, e não apenas com seus valores. Dessa forma, conseguimos acessar um elemento e o elemento seguinte usando lista[j] e lista[j + 1].

A cada passagem, os maiores valores vão sendo deslocados para o final da lista. Esse processo é repetido até que todos os elementos estejam ordenados.

```python
# Cria o tabuleiro com 9 posições vazias
tabuleiro = [" "] * 9

# Define o jogador que começa a partida
jogador = "X"

# Conta quantas jogadas foram realizadas
jogadas = 0


# O jogo pode ter no máximo 9 jogadas,
# pois o tabuleiro possui 9 posições
while jogadas < 9:

    # Mostra o tabuleiro para os jogadores
    print()
    print(tabuleiro[0], "|", tabuleiro[1], "|", tabuleiro[2])
    print("--+---+--")
    print(tabuleiro[3], "|", tabuleiro[4], "|", tabuleiro[5])
    print("--+---+--")
    print(tabuleiro[6], "|", tabuleiro[7], "|", tabuleiro[8])
    print()

    # Solicita ao jogador a posição onde deseja colocar seu símbolo
    # O usuário escolhe de 1 a 9
    posicao = int(input(f"Jogador {jogador}, escolha uma posição de 1 a 9: "))

    # As listas começam no índice 0, por isso diminuímos 1 da posição escolhida
    posicao = posicao - 1

    # Verifica se a posição escolhida já está ocupada
    if tabuleiro[posicao] != " ":
        print("Essa posição já está ocupada!")
        continue

    # Coloca o símbolo do jogador na posição escolhida
    tabuleiro[posicao] = jogador

    # Registra que uma nova jogada foi realizada
    jogadas += 1

    # Verifica todas as possibilidades de vitória:
    # 3 linhas, 3 colunas e 2 diagonais
    if (
        tabuleiro[0] == tabuleiro[1] == tabuleiro[2] != " " or
        tabuleiro[3] == tabuleiro[4] == tabuleiro[5] != " " or
        tabuleiro[6] == tabuleiro[7] == tabuleiro[8] != " " or
        tabuleiro[0] == tabuleiro[3] == tabuleiro[6] != " " or
        tabuleiro[1] == tabuleiro[4] == tabuleiro[7] != " " or
        tabuleiro[2] == tabuleiro[5] == tabuleiro[8] != " " or
        tabuleiro[0] == tabuleiro[4] == tabuleiro[8] != " " or
        tabuleiro[2] == tabuleiro[4] == tabuleiro[6] != " "
    ):
        print(f"Jogador {jogador} venceu!")
        break

    # Alterna o jogador depois de uma jogada válida
    if jogador == "X":
        jogador = "O"
    else:
        jogador = "X"


# Se as 9 posições forem preenchidas sem nenhum vencedor,
# significa que a partida terminou em empate
if jogadas == 9:
    print("Empate!")
```

JUSTIFICATIVA: O tabuleiro foi representado por uma lista com 9 posições, porque o jogo da velha possui 9 espaços. Cada posição da lista corresponde a um espaço do tabuleiro.

Usamos "while jogadas < 9" porque uma partida pode ter no máximo 9 jogadas. Assim, temos uma condição clara para controlar o jogo.

A variável jogador indica quem deve jogar. Depois de cada jogada válida, ela é alterada de "X" para "O" ou de "O" para "X".

As condições de vitória verificam as 8 combinações possíveis: 3 linhas; 3 colunas; 2 diagonais.

O "continue" é utilizado quando o jogador escolhe uma posição que já está ocupada. Nesse caso, o programa volta para o início do "while" sem contar aquela tentativa como uma jogada. O "break" encerra o jogo quando um jogador consegue uma combinação vencedora. Se ninguém vencer e as 9 posições forem preenchidas, o programa informa que houve empate.

35. Números Fibonacci Até N: Gere a sequência de Fibonacci até um número n fornecido pelo
usuário.

```python
# Solicita ao usuário o limite da sequência
n = int(input("Até qual número voce deseja gerar a sequência de Fibonacci? "))

# Os dois primeiros números da sequência de Fibonacci
lista = [0, 1]

# Enquanto o próximo número não ultrapassar o limite informado
while lista[-1] + lista[-2] <= n:

    # O próximo número é obtido somando os dois últimos números
    prox_elemento = lista[-1] + lista[-2]

    # Adiciona o novo número à sequência
    lista.append(prox_elemento)

# Exibe a sequência gerada
print(lista)
```

JUSTIFICATIVA: Foi utilizado um while para continuar gerando novos termos enquanto a soma dos dois últimos elementos da lista não ultrapassar o limite definido pelo usuário. Os índices negativos -1 e -2 permitem acessar, respectivamente, o último e o penúltimo elemento da lista.

A cada repetição, esses dois valores são somados para obter o próximo termo, que é adicionado à lista utilizando append().

36. Gerador de Senhas: Crie um gerador de senhas aleatórias com letras e números.

```python
import random

# Define os caracteres que poderão aparecer na senha (letras maiusculas, minusculas, numeros e alguns simbolos especiais)
caracteres = "ABCDEFGHIJKLMNOPQRSTUWXYZabcdefghijklmnopqrstuvwxyz0123456789!.*_-=+"

# Solicita ao usuário o tamanho desejado para a senha
tamanho = int(input("Qual será o tamanho da senha? "))

# Lista que armazenará os caracteres escolhidos
senha = []

# Repete o processo até atingir o tamanho definido pelo usuário
for i in range(tamanho):

    # Escolhe aleatoriamente um caractere da lista de caracteres disponíveis
    caractere = random.choice(caracteres)

    # Adiciona o caractere escolhido à senha
    senha.append(caractere)

# Junta os caracteres da lista em uma única string
senha = "".join(senha)

# Exibe a senha gerada
print(f"Senha gerada: {senha}")
```

JUSTIFICATIVA: Foi utilizado o módulo "random" para realizar a escolha aleatória dos caracteres que irão formar a senha. A string caracteres contém letras maiúsculas, minúsculas e números, atendendo ao requisito do exercício.

O tamanho da senha é definido pelo usuário e utilizado no for, que repete a escolha de um caractere até atingir a quantidade desejada. Cada caractere escolhido é armazenado em uma lista e ao final o método ".join()" transforma essa lista em uma única string, formando a senha.

O "random.choice()" permite que cada posição da senha receba um caractere escolhido aleatoriamente, fazendo com que diferentes senhas sejam geradas a cada execução.

37. Funções Recursivas: Implemente uma função recursiva para calcular o fatorial de um
número.

```python
# Função que calcula o fatorial de um número
def fatorial(n):

    # O fatorial de 0 e de 1 é igual a 1.
    # Essa é a condição que encerra a recursão.
    if n == 0 or n == 1:
        return 1

    # A função chama ela mesma usando o número anterioR, exemplo: 5! = 5 * 4!
    return n * fatorial(n - 1)


# Solicita um número ao usuário
n = int(input("Digite um número: "))

# Chama a função e exibe o resultado
print(f"O fatorial de {n} é {fatorial(n)}.")
```

JUSTIFICATIVA: Foi utilizada uma função recursiva onde uma função chama a si mesma para resolver o problema.

A condição n == 0 or n == 1 funciona como condição de parada, impedindo que a função continue chamando a si mesma indefinidamente. Quando essa condição é atingida, a função retorna 1, permitindo que os resultados das chamadas anteriores sejam calculados até chegar ao resultado final.

42. Números Primos até N: Gere todos os números primos até um número n fornecido.

```python
from math import sqrt


n = int(input("Insira um valor para gerar os números primos até ele: "))


# Percorremos todos os números de 2 até n.
# Começamos em 2 porque 0 e 1 não são números primos.
for numero in range(2, n + 1):

    # Calculamos a raiz quadrada do número que estamos verificando.
    raiz = sqrt(numero)

    # Inicialmente, consideramos que o número é primo.
    primo = True

    # Percorremos os possíveis divisores do número.
    # Só precisamos verificar até a raiz quadrada.
    for x in range(2, int(raiz) + 1):

        # Se o resto da divisão for 0, encontramos um divisor.
        if numero % x == 0:

            # Portanto, o número não é primo.
            primo = False

            # Não precisamos continuar procurando divisores.
            break

    # Se nenhum divisor foi encontrado, o número é primo.
    if primo:
        print(numero)
```

JUSTIFICATIVA: foi utilizado um for com range(2, n + 1), que percorre todos os números começando em 2 até o valor informado. O +1 é necessário porque o range() não inclui o último valor, portanto, dessa forma, o próprio n também é verificado.

Para cada número percorrido, é calculada sua raiz quadrada utilizando sqrt(). Essa raiz é utilizada para limitar a procura por divisores, pois não é necessário testar números maiores que ela para descobrir se o número é primo.

Um segundo for percorre os possíveis divisores, começando em 2 e indo até a raiz quadrada do número. O operador % verifica se a divisão é exata. Caso o resto seja 0, significa que foi encontrado um divisor e, portanto, o número não é primo. Nesse caso, a variável primo recebe False e o break encerra a verificação daquele número.

46. Calcular Média e Desvio Padrão: Calcule a média e o desvio padrão de uma lista de números.

```python
# Para evitar cálculo manual, foi importada a função sqrt que permite calcular a raiz quadrada de um valor fornecidoV automaticamente.
from math import sqrt

lista = [10,5,20,5]

# A média foi calculada somando todos os valores da lista com o "sum" e dividindo pelo tamanho (len) dessa lista.
media = sum(lista)/len(lista)

# A lista armazenará as diferenças entre o elemento da lista e a média dessa lista (operação necessária para o cálculo do desvio padrão)
diferencas = []


for x in lista:
    dif = x - media # Diferença entre o elemento e a média
    y = dif**2 # Quadrado da diferenca calculada anteriormente
    diferencas.append(y) # Adicionando na lista de diferencas

# O próximo passo para calcular o desvio padrão, é fornecer a média das diferenças:
media_diferencas = sum(diferencas)/len(diferencas)

# Por fim, calculamos a raiz quadrada dessa média das diferenças.
desvio_padrao = sqrt(media_diferencas)

print(f'Média: {media}')
print(f'Desvio padrão: {desvio_padrao:.2f}')

```

JUSTIFICATIVA: O programa tem como objetivo calcular a média e o desvio padrão dos valores de uma lista. Primeiro, a média é obtida somando todos os valores e dividindo pela quantidade de elementos.

Em seguida, para calcular o desvio padrão, o programa percorre cada valor da lista e calcula sua diferença em relação à média. Essas diferenças são elevadas ao quadrado para evitar que valores positivos e negativos se anulem. Os resultados são armazenados em uma nova lista.

Depois, é calculada a média dessas diferenças ao quadrado. Por fim, utiliza-se a função sqrt() para obter a raiz quadrada desse resultado, chegando ao desvio padrão.

A formatação :.2f foi utilizada para apresentar o desvio padrão com duas casas decimais, facilitando a leitura do resultado.

48. Ordenação por Inserção: Implemente o algoritmo de ordenação por inserção (Insertion Sort).

```python
lista = [5, 2, 8, 1, 3]

# Começamos pelo segundo elemento, pois o primeiro
# sozinho já pode ser considerado ordenado
for i in range(1, len(lista)):

    # Guarda o elemento que será colocado na posição correta
    elemento = lista[i]

    # Começa comparando com o elemento anterior
    j = i - 1

    # Enquanto o elemento anterior for maior que o elemento atual,
    # deslocamos o elemento anterior uma posição para a direita
    while j >= 0 and lista[j] > elemento:

        lista[j + 1] = lista[j]

        # Passamos para a posição anterior
        j -= 1

    # Coloca o elemento na posição correta
    lista[j + 1] = elemento

print(lista)
```

JUSTIFICATIVA: O algoritmo recebe uma lista e considera inicialmente o primeiro elemento como ordenado. A partir do segundo elemento, cada valor é comparado com os elementos anteriores. Quando um elemento anterior é maior, ele é deslocado uma posição para a direita. Depois que encontramos a posição correta, o elemento é inserido nela.

O for começa em 1 porque não precisamos comparar o primeiro elemento com ele mesmo. O while é utilizado para continuar voltando pela parte já ordenada enquanto encontrar elementos maiores que o valor que estamos inserindo.

Assim, a parte esquerda da lista vai ficando ordenada progressivamente até que toda a lista esteja organizada.

50. Cálculo de Frequência de Letras: Calcule a frequência de cada letra em uma string fornecida pelo
usuário.

```python
palavra = input("Digite uma palavra: ")

# Transforma a palavra em uma lista, para que seja possivel percorrer cada item
palavra = list(palavra)

# Nesse dicionario "contador", a chave será a letra da palavra que estará presente na lista anterior, e o valor será o número de ocorrências dessa letra.
contador = {}

# Itera cada item (letra)
for letra in palavra:
  if letra in contador:
    # Adiciona +1 caso já exista a letra no contador.
    contador[letra] +=1

  else:
    # Caso ainda não exista, adiciona 1 ao valor.
    contador[letra] = 1

print(contador)
```

JUSTIFICATIVA: O programa tem como objetivo contar quantas vezes cada letra aparece em uma palavra. Para isso, a palavra é transformada em uma lista, permitindo percorrer cada letra individualmente.

Foi utilizado um dicionário como contador, no qual cada letra representa uma chave e seu respectivo valor indica a quantidade de vezes que ela aparece. Durante o for, o programa verifica se a letra já existe no dicionário. Caso exista, seu contador é incrementado em 1; caso contrário, a letra é adicionada ao dicionário com o valor inicial 1.

Dessa forma, ao final da execução, o dicionário apresenta cada letra encontrada e sua quantidade de ocorrências.
