```python
# 2. Crie um programa que peça ao usuário dois números e exiba a soma deles.

a = int(input("Insira um número: "))
b = int(input("Insira outro número: "))

soma = a + b
print(f"A soma entre {a} e {b} é igual a {soma}")

```

```python
# 3. Solicite o raio de um círculo e calcule sua área. Use a fórmula área = π * r².

from math import pi

r = float(input("Insira o valor do raio: "))
area = pi * (r**2)

print(f"A área do círculo é igual a {area:.2f}") # Formatação para retornar com só duas casas decimais
```

```python
# 4. Converta uma temperatura fornecida em Celsius para Fahrenheit usando a fórmula F = C * 9/5 + 32.

c = float(input("Insira o valor (em Celsius) da temperatura: "))
f = c * 9/5 + 32

print(f"Conversão para Fahrenheit: {f}")
```

```python
# 5. Crie um programa que imprima a tabuada de um número fornecido pelo usuário.
n = int(input("Insira um número de 1 a 10: "))
x = 1

while x <= 10: # Enquanto x for menor ou igual a 10, o loop continuará.
    y = n*x
    print(f"{n} x {x} = {y}")
    x += 1
```

```python
# 6. Faça um programa que exiba todos os números pares de 1 a 20.
x = 1
while x <= 20:
  if (x % 2 == 0): # Verifica se o "resto" da divisão por 2 é igual a 0
    print(x)
  x += 1

```

```python
# 7. Solicite um número ao usuário e informe se ele é par ou ímpar.
n = int(input("Informe um valor: "))

if(n % 2 == 0):
  print(f"O número {n} é par.")
else:
  print(f"O número {n} é ímpar.")
```
